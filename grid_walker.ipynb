```python
GRID_SIZE = 10
N_BLACK = 15

numbers = [1,2,3,4,5,6]
movement = {1:"up",2:"up",3:"up",4:"down",5:"right",6:"left"}

rand = random.Random()

all_cells = [(x,y) for x in range(GRID_SIZE) for y in range(GRID_SIZE)]
black_cells = set(rand.sample(all_cells, N_BLACK))

def random_free_on_row(y):
    choices = [ (x,y) for x in range(GRID_SIZE) if (x,y) not in black_cells ]
    return rand.choice(choices)

red = random_free_on_row(0)
blue = random_free_on_row(GRID_SIZE-1)
while blue == red:
    blue = random_free_on_row(GRID_SIZE-1)

plt.ion()
fig, ax = plt.subplots(figsize=(6,6))

steps = 0
while True:
    roll = rand.randint(1,6)
    steps += 1

    dir = movement[roll]
    dx, dy = 0, 0
    if dir == "up":
        dy = 1
    elif dir == "down":
        dy = -1
    elif dir == "right":
        dx = 1
    elif dir == "left":
        dx = -1

    candidate = (red[0] + dx, red[1] + dy)
    if 0 <= candidate[0] < GRID_SIZE and 0 <= candidate[1] < GRID_SIZE:
        if candidate not in black_cells:
            red = candidate

    ax.cla()
    ax.set_xlim(0, GRID_SIZE)
    ax.set_ylim(0, GRID_SIZE)
    ax.set_xticks(range(GRID_SIZE+1))
    ax.set_yticks(range(GRID_SIZE+1))
    ax.grid(True, which='both', color='black', linewidth=1, linestyle='-')
    ax.set_xticklabels([])
    ax.set_yticklabels([])
    ax.set_aspect('equal')

    for (x,y) in black_cells:
        ax.add_patch(patches.Rectangle((x,y),1,1,facecolor='black',edgecolor='none'))

    ax.add_patch(patches.Rectangle(blue,1,1,facecolor='blue',edgecolor='none'))
    ax.add_patch(patches.Rectangle(red,1,1,facecolor='red',edgecolor='none'))

    ax.set_title(f"Roll: {roll}    Steps: {steps}")
    print(f"Roll {steps}: {roll} -> {movement[roll]} | Position: {red}")

    plt.draw()
    plt.pause(2)

    if red == blue:
        print(f"Reached blue in {steps} rolls.")
        break

plt.ioff()
plt.show()
```
